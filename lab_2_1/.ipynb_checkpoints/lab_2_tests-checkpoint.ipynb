```python
import pandas as pd
import numpy as np

def scale(X, x_min, x_max):
    nom = (X-X.min(axis=0))*(x_max-x_min)
    denom = X.max(axis=0) - X.min(axis=0)
    denom[denom==0] = 1
    return x_min + nom/denom

# we import raw_data from the data folder
raw_data_1 = pd.read_csv(
    "data/datatest.txt",
    header=None,
    skiprows=1,
    names=[
        "Index",
        "date",
        "Temperature",
        "Humidity",
        "Light",
        "CO2",
        "HumidityRatio",
        "Occupancy"
    ]
)
```
